```r
# ----------------------------
# 1. sub() - Replace first occurrence
# ----------------------------

y <- "Number of participants: 25"
sub("25", "30", y)


# Example: Replace first occurrence of a name

y <- "Mr. Singh is the smart one. Mr. Singh is funny, too."
sub("Mr. Singh", "Professor Jha", y)

```

```r
# ----------------------------
# 2. gsub() - Replace all occurrences
# ----------------------------

y <- "Mr. Singh is the smart one. Mr. Singh is funny, too."
gsub("Mr. Singh", "Professor Jha", y)


# Compare sub() and gsub()

sub("Mr. Singh", "Professor Jha", y)
gsub("Mr. Singh", "Professor Jha", y)

```

```r

# ----------------------------
# 3. grep() - Search for matching strings
# ----------------------------

str <- c(
  "R Course",
  "exercises",
  "include examples of R language"
)

# Return matching strings
grep("ex", str, value = TRUE)

# Return matching indices
grep("ex", str, value = FALSE)

# Case-insensitive search
grep("R", str, ignore.case = TRUE, value = TRUE)

# Case-sensitive search
grep("R", str, ignore.case = FALSE, value = TRUE)

```

```r

# ----------------------------
# 4. grep() - Search in two strings
# ----------------------------

x <- "R course 24.07.2021"
y <- "Number of participants: 25"

# Combine strings
c(x, y)

# Search for "our"
grep("our", c(x, y))

# Search for "Num"
grep("Num", c(x, y))

```

```r

# ----------------------------
# 5. grepl() - Return TRUE or FALSE
# ----------------------------

str <- c(
  "R Course",
  "exercises",
  "include examples of R language"
)

grepl("R", str)

grepl("ex", str)

```

```r

# ============================================================
# LECTURE 41: DATA FRAMES
# ============================================================

# ----------------------------
# 6. Load MASS package
# ----------------------------

if (!requireNamespace("MASS", quietly = TRUE)) {
  install.packages("MASS")
}

library(MASS)

```

```r

# ----------------------------
# 7. Load painters dataset
# ----------------------------

painters <- MASS::painters

# Display complete dataset
print(painters)

# Display first few rows
head(painters)

# Display last few rows
tail(painters)


```

```r

# ----------------------------
# 8. Display row names
# ----------------------------

rownames(painters)

```

```r

# ----------------------------
# 9. Extract columns
# ----------------------------

painters$Composition
painters$Drawing
painters$Colour
painters$Expression
painters$School

```

```r

# ----------------------------
# 10. Check data types
# ----------------------------

is.numeric(painters$School)
is.numeric(painters$Drawing)

is.factor(painters$School)
is.factor(painters$Drawing)


# ----------------------------
# 11. Display column names
# ----------------------------

colnames(painters)


# ----------------------------
# 12. Summary of complete dataset
# ----------------------------

summary(painters)
```

```r

# ============================================================
# LECTURE 42: DATA FRAMES - CREATION AND OPERATIONS
# ============================================================

# ----------------------------
# 13. Summary of categorical variable
# ----------------------------

summary(painters$School)


# ----------------------------
# 14. Attach a data frame
# ----------------------------

attach(painters)

# Access variables directly
summary(School)
summary(Composition)

# Detach the dataset
detach(painters)

```

```r

# ----------------------------
# 15. Subset by School
# ----------------------------

subset(painters, School == "F")


# Equivalent method using indexing

painters[painters[["School"]] == "F", ]


# ----------------------------
# 16. Subset by Composition
# ----------------------------

subset(painters, Composition <= 6)

```

```r

# ----------------------------
# 17. Select specific columns
# ----------------------------

subset(
  painters,
  School == "F",
  select = c(-3, -5)
)


# ----------------------------
# 18. Split data frame by School
# ----------------------------

splitted <- split(painters, painters$School)

# Display split data
splitted

# Display individual groups
splitted$A
splitted$B
splitted$C
splitted$D
splitted$E
splitted$F
splitted$G
splitted$H


# ----------------------------
# 19. Check whether a split is a data frame
# ----------------------------

is.data.frame(splitted$A)

```

```r

# ============================================================
# LECTURE 44: DATA FRAMES - COMBINING AND MERGING
# ============================================================

# ----------------------------
# 20. Create first data frame
# ----------------------------

df1 <- data.frame(
  state = c("UP", "MP", "AP", "JK"),
  popnsize = c(1000, 2000, 3000, 4000)
)

print(df1)


# ----------------------------
# 21. Create second data frame
# ----------------------------

df2 <- data.frame(
  state = c("UP", "MP", "AP", "JK"),
  samplesize = c(100, 200, 300, 400),
  surveycompleted = c("Yes", "No", "Yes", "No")
)

print(df2)


```

```r

# ----------------------------
# 22. cbind() - Combine columns side by side
# ----------------------------

cbind(df1, df2)


# ----------------------------
# 23. merge() - Join using common column
# ----------------------------

merge(df1, df2, by = "state")
```

```r

# ----------------------------
# 24. rbind() - Combine rows vertically
# ----------------------------

df11 <- data.frame(
  state = c("UP", "MP", "AP", "JK"),
  popnsize = c(1000, 2000, 3000, 4000)
)

df22 <- data.frame(
  state = c("Bihar", "Delhi", "Punjab"),
  popnsize = c(100, 200, 300)
)

# Display data frames
print(df11)
print(df22)

# Stack rows
rbind(df11, df22)


# ============================================================
# END OF LECTURES 40 TO 44
# ============================================================
```
